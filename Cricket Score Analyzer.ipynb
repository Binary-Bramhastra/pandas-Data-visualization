```python
import pandas as pd
```

```python
# 1. Sample Data (Match stats: Player, Runs, Balls Faced, Innings)
data = {
    'Player': ['Dhoni', 'Kohli', 'Gaikwad', 'Sharma', 'Yadav', 'Pandya'],
    'Runs': [200, 100, 150, 100, 50, 30],
    'Balls_Faced': [112, 70, 80, 45, 20, 6],
    'Out': [0, 1, 1, 1, 1, 0] # 1 means out, 0 means not out
}

df = pd.DataFrame(data)
```

```python
# 2. Group data by Player to calculate career/series stats
analyzer = df.groupby('Player').agg({
    'Runs': 'sum',
    'Balls_Faced': 'sum',
    'Out': 'sum'
})
```

```python
# 3. Calculate Batting Average and Strike Rate
# Batting Average = Total Runs / Total Times Out
# We replace 0 with 1 to avoid division by zero (for Not Outs)
analyzer['Average'] = analyzer['Runs'] / analyzer['Out'].replace(0, 1)
```

```python
# 4. Strike Rate = (Total Runs / Total Balls) * 100
analyzer['Strike_Rate'] = (analyzer['Runs'] / analyzer['Balls_Faced']) * 100
```

```python
print("--- Cricket Performance Analysis ---")
print(analyzer.round(2))
```
