```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
# Store initial data
books = pd.DataFrame({
    "Book_ID": [1, 2, 3, 4, 5, 6, 7],
    "Title": ["Python Basics", "Data Science", "Machine Learning", "Deep Learning", "Mathematics","Quantam Physics", "5 vi ki marathi book"],
    "Author": ["John", "David", "Andrew", "Vilson","Shreeansh","Pranjal.d","Tarang"],
    "Category": ["Programming", "Data", "AI", "ML","Maths","Physics","Marathi"],
    "Quantity": [5, 3, 4, 2, 6, 12, 102]
})

members = pd.DataFrame({
    "Member_ID": ["MEM1001", "MEM1002","MEM1003","MEM1004","MEM1005","MEM1006","MEM1007","MEM1008"],
    "Name": ["Pranjal", "Pd", "Gargi", "Anjali", "Tarang", "Shreeansh", "Harshad", "Anuj"],
    "Phone": ["9876543210", "8765432109", "7896543290","5678903412","8906543210","6578904536","9087545678","6789543210"],
    "Email": ["pranjal@gmail.com", "pdeshpande@gmail.com","gargi@gmail.com","anjali@gmail.com","tarang@gmail.com","shreeansh@gmail.com","harshad@gmail.com","anuj@gmail.com"]
})

issued_books = pd.DataFrame(columns=[
    "Issue_ID",
    "Member_ID",
    "Book_ID",
    "Issue_DateTime",
    "Return_DateTime",
    "Status"
])
```

```python
# Book Management

def add_book():
    global books

    book_id = len(books) + 1

    title = input("Binary Bramastra ")
    author = input("Tarang ")
    category = input("all in one")
    quantity = int(input("10"))

    new_book = pd.DataFrame({
        "Book_ID": [book_id],
        "Title": [title],
        "Author": [author],
        "Category": [category],
        "Quantity": [quantity]
    })

    books = pd.concat([books, new_book], ignore_index=True)

    print("Book Added Successfully")


def update_book():
    global books

    book_id = int(input("3 "))

    if book_id not in books["Book_ID"].values:
        print("Book Not Found")
        return

    qty = int(input("7 "))

    books.loc[books["Book_ID"] == book_id, "Quantity"] = qty

    print("Book Updated Successfully")


def view_books():
    print("\n===== BOOKS =====")
    print(books)

```

```python
# Member Management

def register_member():
    global members

    member_id = "MEM" + str(1000 + len(members) + 1)

    name = input("Asavari")
    phone = input("8788974391 ")
    email = input("asavari@gmail.com ")

    new_member = pd.DataFrame({
        "Member_ID": [member_id],
        "Name": [name],
        "Phone": [phone],
        "Email": [email]
    })

    members = pd.concat([members, new_member], ignore_index=True)

    print("Member Registered Successfully")
    print("Generated Member ID:", member_id)


def view_members():
    print("\n===== MEMBERS =====")
    print(members)
```

```python
# Issue book

def issue_book():
    global books, issued_books

    member_id = input("MEM1006 ")
    book_id = int(input(" 2"))

    if member_id not in members["Member_ID"].values:
        print("Invalid Member ID")
        return

    if book_id not in books["Book_ID"].values:
        print("Invalid Book ID")
        return

    qty = books.loc[
        books["Book_ID"] == book_id,
        "Quantity"
    ].values[0]

    if qty <= 0:
        print("Book Not Available")
        return

    books.loc[
        books["Book_ID"] == book_id,
        "Quantity"
    ] -= 1

    issue_id = len(issued_books) + 1

    issue_time = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )

    new_issue = pd.DataFrame({
        "Issue_ID": [issue_id],
        "Member_ID": [member_id],
        "Book_ID": [book_id],
        "Issue_DateTime": [issue_time],
        "Return_DateTime": [""],
        "Status": ["Issued"]
    })

    issued_books = pd.concat(
        [issued_books, new_issue],
        ignore_index=True
    )

    print("Book Issued Successfully")

```

```python
# View Issued Books

def view_issued_books():
    print("\n===== ISSUED BOOKS =====")
    print(issued_books)


# =========================
# Statistics + Visualization
# =========================

def statistics():

    total_books = np.sum(books["Quantity"])
    total_members = len(members)
    total_issued = len(
        issued_books[
            issued_books["Status"] == "Issued"
        ]
    )

    print("\n===== LIBRARY STATISTICS =====")
    print("Total Books Available :", total_books)
    print("Total Members :", total_members)
    print("Currently Issued :", total_issued)

    plt.figure(figsize=(8,5))

    plt.bar(
        books["Title"],
        books["Quantity"],
        color="skyblue"
    )

    plt.title("Book Availability")
    plt.xlabel("Books")
    plt.ylabel("Quantity")

    plt.show()
```

```python
# Main Menu
# =========================

while True:

    print("\n")
    print("=" * 40)
    print("LIBRARY MANAGEMENT SYSTEM")
    print("=" * 40)

    print("1. Add Book")
    print("2. Update Book")
    print("3. View Books")
    print("4. Register Member")
    print("5. View Members")
    print("6. Issue Book")
    print("7. Return Book")
    print("8. View Issued Books")
    print("9. Statistics")
    print("10. Exit")

    choice = input("Enter Choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        update_book()

    elif choice == "3":
        view_books()

    elif choice == "4":
        register_member()

    elif choice == "5":
        view_members()

    elif choice == "6":
        issue_book()

    elif choice == "7":
        return_book()

    elif choice == "8":
        view_issued_books()

    elif choice == "9":
        statistics()

    elif choice == "10":
        print("Thank You")
        break

    else:
        print("Invalid Choice")
```
