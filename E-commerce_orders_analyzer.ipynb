```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create Sample Order Data

data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Customer": ["Pranjal", "Sara", "John", "Priya", "Ali", "Neha", "David", "Riya"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet", "Laptop", "Phone"],
    "Category": ["Electronics", "Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics", "Electronics"],
    "Quantity": [1, 2, 1, 1, 3, 2, 1, 2],
    "Price": [50000, 20000, 15000, 52000, 21000, 14000, 51000, 22000]
}

# Create DataFrame
df = pd.DataFrame(data)
```

```python
 # Add Total Amount Column

df["Total_Amount"] = df["Quantity"] * df["Price"]

# Display Dataset
print("\n===== E-commerce Orders Data =====")
print(df)
```

```python
# Basic Analysis

# Total Revenue
total_revenue = df["Total_Amount"].sum()

print("\n===== Total Revenue =====")
print("₹", total_revenue)

# Average Order Value
average_order = df["Total_Amount"].mean()

print("\n===== Average Order Value =====")
print("₹", round(average_order, 2))

# Top Selling Product
top_product = df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)

print("\n===== Top Selling Products =====")
print(top_product)

# Highest Revenue Product
revenue_product = df.groupby("Product")["Total_Amount"].sum().sort_values(ascending=False)

print("\n===== Product Revenue =====")
print(revenue_product)
```

```python
# NumPy Analysis

amount_array = np.array(df["Total_Amount"])

print("\n===== NumPy Statistics =====")
print("Maximum Order Value:", np.max(amount_array))
print("Minimum Order Value:", np.min(amount_array))
print("Mean Order Value:", np.mean(amount_array))
```

```python
# Data Visualization

sns.set_style("whitegrid")

# 1. Product Sales Chart
plt.figure(figsize=(8, 5))
top_product.plot(kind="bar", color="skyblue")

plt.title("Top Selling Products")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.show()
```

```python
# 2. Revenue by Product
plt.figure(figsize=(8, 5))
revenue_product.plot(kind="pie", autopct='%1.1f%%')

plt.title("Revenue Share by Product")
plt.ylabel("")

plt.show()
```

```python
# 3. Order Amount Distribution
plt.figure(figsize=(8, 5))

sns.histplot(df["Total_Amount"], bins=5, kde=True, color="green")

plt.title("Order Amount Distribution")
plt.xlabel("Total Amount")

plt.show()

print("\n===== Analysis Completed Successfully =====")
```
