```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Create Netflix Dataset

data = {
    "Title": [
        "Stranger Things", "Money Heist", "Wednesday",
        "Dark", "Squid Game", "Narcos",
        "The Witcher", "Breaking Bad", "Lucifer",
        "Extraction"
    ],

    "Type": [
        "TV Show", "TV Show", "TV Show",
        "TV Show", "TV Show", "TV Show",
        "TV Show", "TV Show", "TV Show",
        "Movie"
    ],

    "Release_Year": [
        2016, 2017, 2022,
        2017, 2021, 2015,
        2019, 2008, 2016,
        2020
    ],

    "Rating": [
        "TV-14", "TV-MA", "TV-14",
        "TV-MA", "TV-MA", "TV-MA",
        "TV-14", "TV-MA", "TV-14",
        "R"
    ],

    "Category": [
        "Sci-Fi", "Crime", "Fantasy",
        "Sci-Fi", "Thriller", "Crime",
        "Fantasy", "Crime", "Fantasy",
        "Action"
    ],

    "Duration": [
        50, 45, 40,
        60, 55, 50,
        60, 47, 42,
        120
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Dataset Information

print("Dataset Shape : ")
print(df.shape)

print("\nColumn Names : ")
print(df.columns)

print("\nFirst 5 Rows")
print(df.head(2))
```

```python
# Using NumPy

average_duration = np.mean(df["Duration"])

print("Average Duration :", average_duration)

max_duration = np.max(df["Duration"])

print("Maximum Duration :", max_duration)
```

```python
# Count Types

type_count = df["Type"].value_counts()

print(type_count)
```

```python
# Bar Chart for Content Type

plt.figure(figsize=(6,5))

type_count.plot(kind="bar")

plt.title("Movies vs TV Shows ")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
# Release Year Analysis

year_count = df["Release_Year"].value_counts().sort_index()

print(year_count)
```

```python
# Line Chart

plt.figure(figsize=(8,5))

plt.plot(
    year_count.index,
    year_count.values,
    marker="o"
)

plt.title("Netflix Content Release Trend")
plt.xlabel("Year")
plt.ylabel("Number of Shows")

plt.grid()

plt.show()
```

```python
# Genre Count

genre_count = df["Category"].value_counts()

print(genre_count)
```

```python


# genres ko count karega
genre_data = df["Category"].value_counts()

# figure banayenga 
fig = plt.figure(figsize=(8,6))

#3D graph ko aad karga
ax = fig.add_subplot(111, projection="3d")

# X positions
x = np.arange(len(genre_data))

# Y positions
y = np.zeros(len(genre_data))

# Bar height
z = genre_data.values

#  3D bars ko banaye ga / design karega
ax.bar3d(
    x,          # x-axis
    y,          # y-axis
    np.zeros(len(genre_data)),
    0.5,        # width
    0.5,        # depth
    z           # height
)

# Genre names
#ax.set_xticks(x)
#ax.set_xticklabels(genre_data.index)

# Labels
ax.set_xlabel("Category")
ax.set_zlabel("Count")

# Title
plt.title("Netflix Genre Analysis")

# Show graph
plt.show()
```
