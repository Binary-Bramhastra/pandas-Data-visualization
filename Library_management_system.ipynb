```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Create the library database
data = {
    'Title': ['Python', 'Data Science', 'SQL'],
    'Copies': [10, 5, 8],
    'Borrowed': [2, 1, 4]
}
df = pd.DataFrame(data)

```

```python
# 2. Function to add a book
def add_book(title, copies):
    global df
    new_book = pd.DataFrame({'Title': [title], 'Copies': [copies], 'Borrowed': [0]})

    
    df = pd.concat([df, new_book], ignore_index=True)
    print(f"Added: {title}")
```

```python
# 3. Add a new book
add_book('Machine Learning', 7)
add_book('Cricket Bharti', 20)

add_book('Quantum Physics', 30)
```

```python
# 4. View the library
print(df)
```

```python
# 5. Simple visualization
df.plot(kind='bar', x='Title', y=['Copies', 'Borrowed'])
plt.title("Library Stock")

plt.show()
```
