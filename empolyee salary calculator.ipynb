```python
import pandas as pd
import numpy as np
```

```python
employee_data = {
    "Employee_ID":[200,20,186,175,33,74,191,220,219],

    "Name":[
         "shree","Harshad","Tarang","Gargii",
    "Pranjal_D","PJ","Sakshii",
    "Anuj","Revatii"

    ],

    "Department":[
        "Professor",
        "Assistant Professor",
        "Lab Staff",
        "Admin Staff",
        "Clerk",
        "Professor","Admin Staff",
        "Clerk",
        "Professor"
    ],

    "Basic_Salary":[100000,150,50000,80000,90000,150000,8999,3456,12345]
    
}

df = pd.DataFrame(employee_data)

df
```

```python
df["HRA"] = df["Basic_Salary"] * 0.20

df["Bonus"] = np.where(
    df["Department"].isin(
        ["Professor","Assistant Professor"]
    ),
    df["Basic_Salary"]*0.10,
    df["Basic_Salary"]*0.05
)

df["Tax"] = df["Basic_Salary"]*0.08

df["Final_Salary"] = (
    df["Basic_Salary"]
    + df["HRA"]
    + df["Bonus"]
    - df["Tax"]
)

df
```

```python


print("\nAverage Salary:")
print(df["Final_Salary"].mean())

print("\nHighest Salary:")
print(df["Final_Salary"].max())

print("\nLowest Salary:")
print(df["Final_Salary"].min())
```

```python
department_report = df.groupby(
    "Department"
)["Final_Salary"].mean()

department_report
```

```python
top_employee = df.loc[
    df["Final_Salary"].idxmax()
]

print(top_employee)
```
