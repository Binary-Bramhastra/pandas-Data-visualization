```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# SAMPLE STOCK DATA (PRELOADED)
# ==============================

data = {
    "Date": pd.date_range(start="2024-01-01", periods=15, freq="D"),
    "Open":  [100, 102, 101, 103, 104, 105, 107, 106, 108, 110, 111, 112, 113, 115, 116],
    "High":  [102, 103, 102, 104, 106, 107, 108, 109, 110, 112, 113, 114, 116, 117, 118],
    "Low":   [99, 100, 100, 101, 102, 104, 105, 104, 106, 108, 109, 110, 111, 113, 114],
    "Close": [101, 102, 101, 103, 105, 106, 107, 108, 109, 111, 112, 113, 115, 116, 117],
    "Volume":[1000, 1200, 1100, 1500, 1300, 1600, 1700, 1650, 1800, 1900, 2000, 2100, 2200, 2300, 2400]
}

df = pd.DataFrame(data)
```

```python
# DISPLAY DATA
# ==============================

print("\n📊 STOCK DATA:")
print(df)

# ==============================
# TECHNICAL ANALYSIS
# ==============================

df["Daily_Return"] = df["Close"].pct_change()

df["MA5"] = df["Close"].rolling(window=5).mean()

df["Volatility"] = df["Daily_Return"].rolling(window=3).std()
```

```python
# NUMPY STATISTICS
# ==============================

mean_price = np.mean(df["Close"])
max_price = np.max(df["Close"])
min_price = np.min(df["Close"])

print("\n📈 STATISTICS:")
print("Mean Close Price:", mean_price)
print("Max Close Price:", max_price)
print("Min Close Price:", min_price)

```

```python
# PLOTS
# ==============================

plt.figure(figsize=(12,6))

plt.plot(df["Date"], df["Close"], label="Close Price", marker="o")
plt.plot(df["Date"], df["MA5"], label="MA 5 Days", linestyle="--")

plt.title("Stock Price Analysis (Offline Sample Data)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# VOLUME ANALYSIS
# ==============================

plt.figure(figsize=(10,5))
plt.bar(df["Date"], df["Volume"], color="orange")
plt.title("Trading Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# DAILY RETURN
# ==============================

plt.figure(figsize=(10,5))
plt.plot(df["Date"], df["Daily_Return"], color="purple", marker="o")
plt.title("Daily Returns")
plt.xticks(rotation=45)
plt.axhline(0, color="black")
plt.tight_layout()
plt.show()

# ==============================
# SAVE DATA
# ==============================

df.to_csv("offline_stock_analysis.csv", index=False)
print("\n✅ Data saved as offline_stock_analysis.csv")
```
